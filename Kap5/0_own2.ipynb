```python
import numpy as np
import pandas as pd
from pandas import Series, DataFrame
import matplotlib.pyplot as plt
from matplotlib import rcParams
import seaborn as sb
```

```python
rcParams['figure.figsize'] = 15,3
sb.set_style('whitegrid')
```

```python
address = 'mtcars.csv'
cars = pd.read_csv(address)
DF_obj = DataFrame(cars)
DF_obj.head()
```

```python
x = range(1, 11)
y = [0,2,3,1,4,7,1,2,5,2]
plt.bar(x,y)
```

```python
DF_obj.plot(kind='bar', x='model', y='gear', color='blue')
```

```python
df_sub = DF_obj[['model', 'gear', 'cyl', 'hp']]
df_sub.plot(
    kind='bar',
    x='model',
    y=['gear', 'cyl',]
)
```
